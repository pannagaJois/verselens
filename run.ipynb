```python
from inference import analyze
from IPython.display import display, Markdown
from interpreter import interpret_poem

print('VerseLens is ready.')
```

```python
poem = """
LORD of the lotus, lord of the harvest,
Bright and munificent lord of the morn!
Thine is the bounty that prospered our sowing,
Thine is the bounty that nurtured our corn.
We bring thee our songs and our garlands for tribute,
The gold of our fields and the gold of our fruit;
O giver of mellowing radiance, we hail thee,
We praise thee, O Surya, with cymbal and flute.

Lord of the rainbow, lord of the harvest,
Great and beneficent lord of the main!
Thine is the mercy that cherished our furrows,
Thine is the mercy that fostered our grain.
We bring thee our thanks and our garlands for tribute,
The wealth of our valleys, new-garnered and ripe;
O sender of rain and the dewfall, we hail thee,
We praise thee, Varuna, with cymbal and pipe.
"""
title="Harvest Hymn: Mens Voices:"
poet="Sarojini Naidu"
if not poem:
    print('Please enter a poem and run this cell again.')

else:
    results = analyze(poem, k=5, threshold=0.3)

    display(Markdown("## VerseLens Results"))

    # 1. Theme and emotion predictions
    for task, predictions in results.items():
        display(Markdown(f"### {task.title()}"))

        if predictions:
            for label, score in predictions:
                print(f"- {label}: {score:.4f}")
        else:
            print("No labels met the threshold.")

    # 2. Detailed literary interpretation using Qwen
    display(Markdown("## Literary Interpretation"))

    interpretation = interpret_poem(
        poem,
        title=title if "title" in locals() else "",
        poet=poet if "poet" in locals() else "",
        max_new_tokens=900,
    )

    display(Markdown(interpretation))
```
